# Launch the Streamlit Demo App from Colab

Runs `app/streamlit_app.py` (the interactive "pick a sentence, see its features, get a live
prediction" demo) inside Colab and exposes it through a public tunnel, since Colab doesn't let
you open a browser port directly.

**This notebook is Colab-only.** Running the app locally doesn't need any of this -- just:
```
cd E:\hindi_wordorder_replication
streamlit run app\streamlit_app.py
```

**Before running this notebook:** upload/sync `hindi_wordorder_replication` to your Google Drive
root (same folder `live_demo_mtp1_mtp2.ipynb` uses) -- this notebook expects `app/streamlit_app.py`,
`data/processed/features/features.csv`, and `hindi_dialog_generation/` to already be there.

**No ngrok account needed** -- this uses `localtunnel`, which is free and anonymous (no signup).
The one quirk: the tunnel shows a "Tunnel Password" gate the first time you open it in a browser;
Cell 4 below prints the password (it's just your Colab VM's public IP) -- copy that value in when asked.

````python
# Start Streamlit as a background process, logging to app_log.txt so you can
# check it if something goes wrong (e.g. a missing package or a data-path issue).
import subprocess
streamlit_proc = subprocess.Popen(
    ['streamlit', 'run', 'app/streamlit_app.py',
     '--server.port', '8501', '--server.headless', 'true'],
    stdout=open('app_log.txt', 'w'), stderr=subprocess.STDOUT,
)
import time
time.sleep(8)
print('Streamlit started (PID', streamlit_proc.pid, ') -- log: app_log.txt')
!tail -n 20 app_log.txt
````

````python
# Your tunnel password (Colab VM's public IP) -- localtunnel will ask for this
# the first time you open the URL printed by the next cell.
!wget -q -O - ipv4.icanhazip.com
````
